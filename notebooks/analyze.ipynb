```python
import numpy as np
import prettypyplot as pplt
from matplotlib import pyplot as plt

from pytrale import Trale

pplt.use_style(figsize=(8, 2), latex=False)
```

```python
# db = Trale.fromFile('../../trale/python/trale_2024-01-29.txt')
N = 150
tmin, tmax = 19000, 19300
times_measured = np.unique(np.random.uniform(tmin, tmax, size=N).astype(int))
N = len(times_measured)
weights_measured = 70 + np.cos(
    times_measured * np.pi / 100
) + np.random.normal(scale=0.3, size=N) - (times_measured - tmin) / 70
db = Trale(
    times_measured=times_measured,
    weights_measured=weights_measured,
)
```

```python
fig, ax = plt.subplots()

ax.scatter(db.times_measured, db.weights_measured)
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(db.times, db.weights_linear_interpol, alpha=0.5)
ax.plot(db.times, db.weights_gaussian_extrapolated, alpha=0.5)
ax.scatter(db.times_measured, db.weights_measured)
plt.show()
```
